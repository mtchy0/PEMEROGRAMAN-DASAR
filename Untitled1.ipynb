```python
#Latihan

# Soal 1 — Kalkulator faktorial 1 sampai 5
n = int(input("Masukkan bilangan bulat (1-5): "))

# Validasi input: n kurang dari 1 ATAU lebih dari 5
if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")

# Cabang untuk n == 1
elif n == 1:
    hasil = 1
    print(f"{n}! = {hasil}")
elif n == 2:
    hasil = 2 * 1
    print(f"{n}! = {hasil}")
elif n == 3:
    hasil = 3 * 2 * 1
    print(f"{n}! = {hasil}")
elif n == 4:
    hasil = 4 * 3 * 2 * 1
    print(f"{n}! = {hasil}")
else:
    hasil = 5 * 4 * 3 * 2 * 1
    print(f"{n}! = {hasil}")
```
