```python
#Latihan

# Input
daya_gen     = float(input("Daya generator (kW)        : "))
jam_operasi  = float(input("Jam operasi per hari       : "))
konsumsi     = float(input("Konsumsi BBM (liter/kWh)   : "))
harga_solar  = float(input("Harga solar (Rp/liter)     : "))
kapasitas    = int(input("Kapasitas tangki (liter)    : "))
arus         = float(input("Arus kabel (A)             : "))
hambatan     = float(input("Hambatan kabel (ohm)       : "))

# 1. Energi per hari
energi = daya_gen * jam_operasi

# 2. Kebutuhan solar dan biaya per hari
solar_hari = energi * konsumsi
biaya      = solar_hari * harga_solar

# 3. Tangki cukup berapa hari penuh, sisa solar
hari_penuh = int(kapasitas // solar_hari)
sisa_solar = kapasitas % solar_hari

# 4. Rugi daya kabel
rugi = arus ** 2 * hambatan

# 5. Apakah rugi kabel tinggi
tinggi = rugi > 150 and arus >= 20

print(f"Energi per hari    : {energi:.0f} kWh")
print(f"Solar per hari     : {solar_hari:.0f} liter")
print(f"Biaya per hari     : Rp {biaya:,.0f}")
print(f"Tangki cukup       : {hari_penuh} hari penuh")
print(f"Sisa solar         : {sisa_solar:.0f} liter")
print(f"Rugi daya kabel    : {rugi:.0f} W")
print(f"Rugi kabel tinggi? : {tinggi}")
```
