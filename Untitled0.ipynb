```python
#Latihan

# Input
jumlah_panel = int(input("Jumlah panel            : "))
daya_panel   = float(input("Daya per panel (Wp)    : "))
jam_sinar    = float(input("Penyinaran efektif (jam/hari): "))
efisiensi    = float(input("Efisiensi sistem (0-1) : "))
kebutuhan    = float(input("Kebutuhan listrik (kWh/hari): "))
tarif        = float(input("Tarif listrik (Rp/kWh) : "))
hari         = 30

# 1. Energi per hari (kWh)
energi_hari = jumlah_panel * daya_panel * jam_sinar * efisiensi / 1000

# 2. Energi per bulan dan nilai rupiah
energi_bulan = energi_hari * hari
nilai_rupiah = energi_bulan * tarif

# 3. Kekurangan energi per hari
kekurangan = kebutuhan - energi_hari

# 4. Apakah mencukupi
cukup = energi_hari >= kebutuhan

print(f"Energi per hari   : {energi_hari:.2f} kWh")
print(f"Energi per bulan  : {energi_bulan:.2f} kWh")
print(f"Nilai rupiah      : Rp {nilai_rupiah:,.2f}")
print(f"Kekurangan/hari   : {kekurangan:.2f} kWh")
print(f"PLTS mencukupi?   : {cukup}")
```
