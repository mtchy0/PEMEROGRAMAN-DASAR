```python
#nama:Mutiara Nursita Siswanto Putri
#nim:4232511051
#tanggal:13 November 2026
#mata kuliah:Pemrograma Dasar

#soal 6

def status_boiler(suhu, tekanan):
    if suhu >= 550 or tekanan >= 14:
        return "STATUS: TRIP"
    elif suhu > 520 and tekanan > 12:
        return "STATUS: WARNING"
    else:
        return "STATUS: NORMAL"

suhu = float(input("Masukkan suhu uap (C): "))
tekanan = float(input("Masukkan tekanan (bar): "))
print(status_boiler(suhu, tekanan))
```

```python
#nama:Mutiara Nursita Siswanto Putri
#nim:4232511051
#tanggal:13 November 2026
#mata kuliah:Pemrograma Dasar

#soal 10

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]
jumlah_valid = 0

for n, teks in enumerate(data, start=1):
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Titik {n}: data tidak valid - pemeriksaan dihentikan")
        break

    if nilai > 70.0:
        print(f"Warning: Titik {n} suhu bearing tinggi!")

    jumlah_valid += 1

print(f"Jumlah data valid: {jumlah_valid}")
```
