```python
#6f

scores = [80, 65, 90, 75, 70]

total = 0
count_pass = 0

for score in scores:
    total = total + score

    if score >= 75:
        count_pass += 1

    print("Running total:", total)

average = total / len(scores)

print("Total:", total)
print("Average:", average)
print("Scores >= 75:", count_pass)
```

```python
#6g

temps = [28, 31, 30, 33, 29, 27, 32]

avg5 = []

for i in range(4, len(temps)):
    window = temps[i-4:i+1]
    avg5.append(round(sum(window) / 5, 1))

print("5-day moving average:", avg5)
print("Number of averages:", len(avg5))
```

```python
#6h

energy = [
    5576, 6098, 6078, 6095, 6074, 6077,
    5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097,
    3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065,
    6071, 6086, 5577, 5575, 6062
]

run_hours = [24] * 30
run_hours[16] = 18

print("Day 17 energy:", energy[16], "MWh")
print("Day 17 operating hours:", run_hours[16])

change = (energy[16] - energy[15]) / energy[15] * 100

print("Energy reduction:", round(change, 2), "%")
```

```python
#6i

temps = [28, 30, 27, 32, 29, 31, 26]

# 1. First 3 days and last 2 days
print("First 3 days:", temps[:3])
print("Last 2 days:", temps[-2:])

# 2. Append and pop
temps.append(33)
print("After append:", temps)

temps.pop()
print("After pop:", temps)

# 3. Sorting
print("Sorted temperatures:", sorted(temps))

# 4. Average and highest
average = sum(temps) / len(temps)
highest = max(temps)
day = temps.index(highest) + 1

print("Average:", round(average, 2))
print("Highest temperature:", highest)
print("Hottest day:", day)

# 5. Temperatures above average
above_average = []

for t in temps:
    if t > average:
        above_average.append(t)

print("Above average:", above_average)
```
