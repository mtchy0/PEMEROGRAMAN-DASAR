```python
#Praktek2.3

def thermal_efficiency(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Thermal efficiency in %, calculated from coal burned (t) and energy sent out (MWh)."""
    heat_rate = coal_t * gcv_kcal_kg / energy_mwh
    return 860 / heat_rate * 100

# Hand calculation: 1 t x 4400 kcal/kg / 2 MWh = 2200 kcal/kWh -> 860 / 2200 = 39.09 %
assert round(thermal_efficiency(1, 2), 2) == 39.09
print("Test passed")

e1 = thermal_efficiency(2998, 5576)                       # correct order
e2 = thermal_efficiency(5576, 2998)                       # swapped!
e3 = thermal_efficiency(energy_mwh=5576, coal_t=2998)     # keywords
for e in [e1, e2, e3]:
    print(round(e, 2), "%", "plausible" if 25 <= e <= 45 else "NOT plausible")
```

```python
#Praktek3.1

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh): heat from the coal needed for each kWh sent out."""
    if energy_mwh <= 0:                                  # stop early if the input makes no sense
        raise ValueError("energy_mwh must be positive")
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate (1 kWh = 860 kcal)."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) over a period."""
    return energy_mwh / (installed_mw * hours) * 100


def specific_coal_consumption(coal_t, energy_mwh):
    """Coal needed per MWh of energy (t/MWh)."""
    return coal_t / energy_mwh


# One test per function, each with a simple number I can calculate by hand
assert heat_rate(1, 2) == 2200                          # 1 x 4400 / 2
assert round(thermal_efficiency(2150), 2) == 40         # 860 / 2150 = 0.4
assert round(capacity_factor(7920)) == 100              # 330 MW x 24 h = 7920 MWh
assert specific_coal_consumption(100, 200) == 0.5       # 100 / 200
print("All four tests passed")

hr = heat_rate(3258, 6098)
print(f"Day 2: HR {hr:,.0f} kcal/kWh, efficiency {thermal_efficiency(hr):.2f} %, "
      f"CF {capacity_factor(6098):.1f} %, SCC {specific_coal_consumption(3258, 6098):.3f} t/MWh")
```

```python
#Praktek3.2

def efficiency_status(efficiency):
    """Give a label for the efficiency: POOR below 36.0, LOW below 36.3, else NORMAL."""
    if efficiency < 36.0:
        return "POOR"          # return ends the function here
    if efficiency < 36.3:
        return "LOW"           # only reached when efficiency >= 36.0
    return "NORMAL"            # only reached when efficiency >= 36.3


# Test exactly on the borders, because that is where < and <= mistakes show up
for e, expected in [(35.9, "POOR"), (36.0, "LOW"), (36.29, "LOW"), (36.3, "NORMAL"), (36.5, "NORMAL")]:
    assert efficiency_status(e) == expected, f"{e} should be {expected}"
    print(f"{e:>5} % -> {efficiency_status(e)}")
print("All limits correct")
```

```python
#Praktek3.3

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) over a period."""
    return energy_mwh / (installed_mw * hours) * 100


def efficiency_status(efficiency):
    """POOR below 36.0 %, LOW below 36.3 %, otherwise NORMAL."""
    if efficiency < 36.0:
        return "POOR"
    if efficiency < 36.3:
        return "LOW"
    return "NORMAL"


def print_row(day, hr, eff, cf, status):
    """Procedure: prints one table row, returns nothing."""
    print(f"{day:>3}  {hr:>7,.0f}  {eff:>6.2f}  {cf:>5.1f}  {status}")


# --- Data Unit 1, 14-20 November 2026 ---
first_day = 14
energy = [5533, 5566, 6097, 3592, 6059, 6100, 6098]      # MWh
coal = [2996, 3009, 3257, 1950, 3234, 3261, 3262]        # t

# --- Main program ---
print(f"{'Day':>3}  {'HR kcal':>7}  {'Eff %':>6}  {'CF %':>5}  Status")
efficiencies = []
for i in range(len(energy)):
    hr = heat_rate(coal[i], energy[i])
    eff = thermal_efficiency(hr)
    efficiencies.append(eff)                              # keep for the sum
    print_row(first_day + i, hr, eff, capacity_factor(energy[i]), efficiency_status(eff))

lowest = min(efficiencies)
print(f"\nAverage efficiency : {sum(efficiencies) / len(efficiencies):.2f} %")
print(f"Lowest efficiency  : {lowest:.2f} % (day {first_day + efficiencies.index(lowest)})")
print(f"Weekly CF          : {capacity_factor(sum(energy), hours=24 * len(energy)):.1f} %")
```

```python
#Praktek3.4

def heat_rate(coal_t, energy_mwh, gcv_kcal_kg=4400):
    """Heat rate (kcal/kWh)."""
    return coal_t * gcv_kcal_kg / energy_mwh


def thermal_efficiency(heat_rate_kcal_kwh):
    """Thermal efficiency (%) from the heat rate."""
    return 860 / heat_rate_kcal_kwh * 100


def capacity_factor(energy_mwh, installed_mw=330, hours=24):
    """Capacity factor (%) over a period."""
    return energy_mwh / (installed_mw * hours) * 100


def unit_report(name, installed_mw, gcv, energy, coal):
    """Procedure: prints the KPI summary of one unit from its energy (MWh) and coal (t) lists."""
    total_energy = sum(energy)
    hr = heat_rate(sum(coal), total_energy, gcv_kcal_kg=gcv)    # total coal / total energy
    cf = capacity_factor(total_energy, installed_mw=installed_mw, hours=24 * len(energy))
    print(f"{name}: {len(energy)} days, HR {hr:,.0f} kcal/kWh, "
          f"efficiency {thermal_efficiency(hr):.2f} %, CF {cf:.1f} %")


# Same functions, only the arguments change from unit to unit
unit_report("Unit 1", 330, 4400,
            [5576, 6098, 6078, 6095, 6074, 6077, 5564],
            [2998, 3258, 3250, 3263, 3249, 3246, 2993])
unit_report("Unit 3", 100, 4200,
            [2210, 2235, 2190],
            [1270, 1281, 1262])

# My own unit: 200 MW, 4300 kcal/kg, 3 days of data
unit_report("Unit 4", 200, 4300,
            [4000, 4100, 3950],
            [2300, 2350, 2270])
```

```python
#Praktek4.exc1

def celsius_to_fahrenheit(celsius):
    """Convert °C to °F: F = C * 9 / 5 + 32."""
    return celsius * 9 / 5 + 32          # return the value, do not print it


def rectangle_area(length, width):
    """Area of a rectangle = length x width."""
    return length * width


def greeting(name):
    """Return the text 'Hello, <name>!' (do not print it)."""
    return f"Hello, {name}!"


# ---- Checker (do not change) ----
def check(text, result, expected):
    try:
        ok = result == expected if isinstance(expected, str) else abs(result - expected) < 0.001
    except TypeError:
        ok = False
    print(("✓ " if ok else "✗ ") + f"{text} = {result!r}" + ("" if ok else f"  (expected {expected!r})"))

check("celsius_to_fahrenheit(100)", celsius_to_fahrenheit(100), 212.0)
check("celsius_to_fahrenheit(-40)", celsius_to_fahrenheit(-40), -40.0)
check("rectangle_area(5, 3)", rectangle_area(5, 3), 15)
check("rectangle_area(width=2, length=10)", rectangle_area(width=2, length=10), 20)
check("greeting('Rina')", greeting("Rina"), "Hello, Rina!")
```

```python
#Praktek4.exc3

def load_percent(load_mw, installed_mw=330):
    """Current load as a percentage of the installed capacity."""
    return load_mw / installed_mw * 100


# ---- Checker ----
def check(text, result, expected):
    ok = result is not None and abs(result - expected) < 0.001
    print(("✓ " if ok else "✗ ") + f"{text} = {result}" + ("" if ok else f"  (expected {expected})"))

check("load_percent(165)", load_percent(165), 50.0)
check("load_percent(330)", load_percent(330), 100.0)
check("load_percent(300, 600)", load_percent(300, 600), 50.0)
check("load_percent(load_mw=450, installed_mw=600)", load_percent(load_mw=450, installed_mw=600), 75.0)
```

```python
#Praktek4.exc4

def load_category(percent):
    """'LOW' if below 50 %, 'HIGH' if above 90 %, otherwise 'NORMAL'."""
    if percent < 50:
        return "LOW"
    if percent > 90:
        return "HIGH"
    return "NORMAL"


# ---- Checker ----
for percent, expected in [(35, "LOW"), (50, "NORMAL"), (76.9, "NORMAL"), (90, "NORMAL"), (95.2, "HIGH")]:
    result = load_category(percent)
    print(("✓ " if result == expected else "✗ ") + f"load_category({percent}) = {result!r}" + ("" if result == expected else f"  (expected {expected!r})"))
```

```python
#Praktek4.exc5

def value_range(values):
    """Return three numbers: lowest, highest, and the difference between them."""
    lowest = min(values)
    highest = max(values)
    return lowest, highest, highest - lowest


# ---- Checker ----
for data, expected in [([5576, 6098, 3592, 6135], (3592, 6135, 2543)), ([36.35, 36.58, 36.0], (36.0, 36.58, 0.58))]:
    result = value_range(data)
    ok = all(abs(r - e) < 0.001 for r, e in zip(result, expected))
    print(("✓ " if ok else "✗ ") + f"value_range({data}) = {result}" + ("" if ok else f"  (expected {expected})"))
```

```python
#Praktek4.exc6

def lowest_days(values, n=3):
    """Day numbers (day 1 is the first value) of the n lowest values, lowest first."""
    copy = values[:]
    days = []
    for _ in range(n):
        i = copy.index(min(copy))
        days.append(i + 1)
        copy[i] = 1e9
    return days


# ---- Checker ----
efficiency = [36.35, 36.58, 36.55, 36.51, 36.54, 36.59, 36.34, 36.38, 36.6, 36.49,
              36.63, 36.63, 36.55, 36.1, 36.15, 36.59, 36.0, 36.62, 36.56, 36.54,
              36.08, 36.36, 36.59, 36.61, 36.5, 36.35, 36.41, 36.36, 36.23, 36.57]
for args, expected in [((efficiency,), [17, 21, 14]), ((efficiency, 1), [17]), (([5.0, 3.0, 4.0], 2), [2, 3])]:
    result = lowest_days(*args)
    print(("✓ " if result == expected else "✗ ") + f"lowest_days(..., n={len(expected)}) = {result}" + ("" if result == expected else f"  (expected {expected})"))
print("Original array unchanged:", "✓" if efficiency[16] == 36.0 else "✗ (do not modify the input array)")
```
